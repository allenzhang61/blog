## SimpleTokenizerV1

```python
import re

class SimpleTokenizerV1:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {v: k for k, v in self.str_to_int.items()}

    def encode(self, text):
        preprocessed = re.split(r'([,.?_!"()\']|--|\s)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        ids = [self.str_to_int[item] for item in preprocessed]
        return ids

    def decode(self, ids):
        text = " ".join([self.int_to_str[id] for id in ids])
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
        return text
```

```python
with open("./data/the-verdict.txt", "r", encoding="utf-8") as f:
        raw_text = f.read()

preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', raw_text)
preprocessed = [item.strip() for item in preprocessed if item.strip()]
```

```python
all_words = sorted(set(preprocessed))
vocab = {token: integer for integer, token in enumerate(all_words)}

print(len(vocab))
```

```python
tokenizer = SimpleTokenizerV1(vocab)
text = """"It's the last he painted, you know," Mrs. Gisburn said with pardonable pride."""
ids = tokenizer.encode(text)
print(ids)

print(tokenizer.decode(ids))
```

## SimpleTokenizerV2

```python
all_tokens = sorted(list(set(preprocessed)))
all_tokens.extend(["<|endoftext|>", "<|unk|>"])
vocab = {token:integer for integer,token in enumerate(all_tokens)}

for i, item in enumerate(list(vocab.items())[-5:]):
      print(item)
```

```python
import re

class SimpleTokenizerV2:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {v: k for k, v in self.str_to_int.items()}

    def encode(self, text):
        preprocessed = re.split(r'([,.?_!"()\']|--|\s)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        # 使用|unk|替换未知词汇
        preprocessed = [item if item in self.str_to_int else '<|unk|>' for item in preprocessed]
        ids = [self.str_to_int[item] for item in preprocessed]
        return ids

    def decode(self, ids):
        text = " ".join([self.int_to_str[id] for id in ids])
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
        return text
```

```python
tokenizer = SimpleTokenizerV2(vocab)
# Hello在 vocab 中不存在
ids = tokenizer.encode(text = """Hello "It's the last he painted, you know," Mrs. Gisburn said with pardonable pride.<|endoftext|>""")
print(ids)

print(tokenizer.decode(ids))
```

```python
import tiktoken
tokenizer = tiktoken.get_encoding('gpt2')
```

```python
text = """Hello someunknownPlace "It's the last he painted, you know," Mrs. Gisburn said with pardonable pride.<|endoftext|>"""
ids = tokenizer.encode(text, allowed_special={"<|endoftext|>"})
print(ids)

print(tokenizer.decode(ids))
for item in ids[:10]:
    print(item, ' -> ',tokenizer.decode([item]))
```
