# 三层神经网络的前向传播

本例实现一个 **2 → 3 → 2 → 2** 的全连接网络，输入 `x = [1.0, 0.5]`。

## 网络结构

| 层 | 变量 | 节点数 | 来源 |
|----|------|--------|------|
| 输入层（第 0 层） | `x` | 2 | `x = [1.0, 0.5]` |
| 隐藏层 1（第 1 层） | `a1` → `z1` | 3 | `W1` 形状 2×3，`b1` 长度 3 |
| 隐藏层 2（第 2 层） | `a2` → `z2` | 2 | `W2` 形状 3×2，`b2` 长度 2 |
| 输出层（第 3 层） | `a3` → `y` | 2 | `W3` 形状 2×2，`b3` 长度 2 |

## 符号说明

- **`a`（加权和 / pre-activation）**：`a = z·W + b`，激活前的值。
- **`z`（激活输出）**：`z = sigmoid(a)`，作为下一层的输入。
- **`W_n` / `b_n`**：下标 `n` 表示「第 n-1 层 → 第 n 层」之间的权重与偏置。
- **输出层激活**：用恒等函数 `identity_function`（回归任务常用；分类任务通常换成 softmax）。

## 前向传播流程

$$
x \xrightarrow{W_1,\,b_1} a_1 \xrightarrow{\text{sigmoid}} z_1
\xrightarrow{W_2,\,b_2} a_2 \xrightarrow{\text{sigmoid}} z_2
\xrightarrow{W_3,\,b_3} a_3 \xrightarrow{\text{identity}} y
$$


```python
import numpy as np
```

```python
def sigmoid(x):
    return 1 / (1+np.exp(-x))

def identity_function(x):
    return x
```

```python
def init_network():
    network = {}
    network['W1'] = np.array([[0.1,0.3,0.5],[0.2,0.4,0.6]])
    network['b1'] = np.array([0.1,0.2,0.3])
    network['W2'] = np.array([[0.1,0.4],[0.2,0.5],[0.3,0.6]])
    network['b2'] = np.array([0.1,0.2])
    network['W3'] = np.array([[0.1,0.3],[0.2,0.4]])
    network['b3'] = np.array([0.1,0.2])
    return network

def forward(network, x):
    W1, W2, W3 = network['W1'], network['W2'], network['W3']
    b1,b2,b3 = network['b1'], network['b2'], network['b3']

    a1 = np.dot(x, W1)+b1
    z1 = sigmoid(a1)
    a2 = np.dot(z1, W2)+b2
    z2 = sigmoid(a2)
    a3 = np.dot(z2, W3)+b3
    y=identity_function(a3)
    return y

network = init_network()
x = np.array([1.0,0.5])
y = forward(network, x)
print(y)
```
