## 为什么线性回归要用 MSE，而不是其他损失函数？

MSE 不是拍脑袋选的，它有严格的概率论依据：**在高斯噪声假设下，最小化 MSE 等价于极大似然估计。**

---

### 第一步：假设噪声服从高斯分布

真实观测值 = 线性预测 + 随机噪声：

$$y = \mathbf{w}^\top\mathbf{x} + b + \epsilon, \quad \epsilon \sim \mathcal{N}(0, \sigma^2)$$

由此，给定输入 $\mathbf{x}$，$y$ 的条件概率也服从高斯分布（均值为预测值）：

$$P(y \mid \mathbf{x}) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\!\left(-\frac{(y - \mathbf{w}^\top\mathbf{x} - b)^2}{2\sigma^2}\right)$$

---

### 第二步：极大似然估计

极大似然估计的目标：找到使"观测数据出现概率最大"的参数 $\mathbf{w}, b$。

对 $n$ 个独立样本，联合似然为各样本概率之积：

$$P(\mathbf{y} \mid \mathbf{X}) = \prod_{i=1}^{n} P(y^{(i)} \mid \mathbf{x}^{(i)})$$

---

### 第三步：取负对数似然，乘积变求和

直接最大化乘积在数值上容易下溢，通常改为最小化**负对数似然**（两者等价）：

$$-\log P(\mathbf{y} \mid \mathbf{X})
= \sum_{i=1}^{n} \frac{1}{2}\log(2\pi\sigma^2)
+ \frac{1}{2\sigma^2}\left(y^{(i)} - \mathbf{w}^\top\mathbf{x}^{(i)} - b\right)^2$$

- **第一项** $\dfrac{n}{2}\log(2\pi\sigma^2)$：与 $\mathbf{w}, b$ 无关，是常数，优化时可忽略
- **第二项** $\dfrac{1}{2\sigma^2}\sum_i\left(y^{(i)} - \hat{y}^{(i)}\right)^2$：$\sigma^2$ 为常数，本质就是 **MSE**

---

### 结论

$$\boxed{\text{最小化 MSE} \iff \text{最大化高斯噪声假设下的似然}}$$

| 噪声假设 | 等价损失函数 |
|---------|------------|
| 高斯分布 $\mathcal{N}(0, \sigma^2)$ | MSE（均方误差）|
| 拉普拉斯分布 $\text{Laplace}(0, b)$ | MAE（平均绝对误差）|

这也解释了为何在有**异常值**（重尾噪声）的场景下，MAE 比 MSE 更鲁棒——因为异常值更符合拉普拉斯分布的假设，而非高斯分布。
