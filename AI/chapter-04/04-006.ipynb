## 线性神经网络和线性回归关系

线性神经网络在最简单的情况下，本质上就是线性回归

### 线性回归问题

对于 1 个样本

线性回归公式：$y=w_1x_1+w_2x_2+\cdots+w_nx_n+b$

记为

$$
y=w^Tx+b=
\begin{bmatrix}
w_1 & w_2 & \cdots & w_n
\end{bmatrix}
\begin{bmatrix}
x_1\\
x_2\\
\vdots\\
x_n
\end{bmatrix}+b
$$

对于多个样本 ($m$个样本，$n$个特征)

$$
\hat y = Xw+b=
\begin{bmatrix}
x_{11} & x_{12} & \cdots & x_{1n}\\
x_{21} & x_{22} & \cdots & x_{2n}\\
\vdots & \vdots & \ddots & \vdots\\
x_{m1} & x_{m2} & \cdots & x_{mn}
\end{bmatrix}
\begin{bmatrix}
w_1\\
w_2\\
\vdots\\
w_n
\end{bmatrix}+
\begin{bmatrix}
b_1\\
b_2\\
\vdots\\
b_n
\end{bmatrix}
$$

### 线性神经网络

深度学习中的单层神经网络：输入层->Linear层->输出层

表示为$Y=XW^T+b$，其中 输入$X: (batch, n)$，权重$W: (out\_features, n)$

当输出特征数为 1 时(out_features=1)，和线性回归完全等价

当输出特征数为 2 时(out_features=2)，可以认为是两个线性回归问题

$$
Y=XW^T+b
=\begin{bmatrix}
x_{11} & x_{12} & \cdots & x_{1n}\\
x_{21} & x_{22} & \cdots & x_{2n}\\
\vdots & \vdots & \ddots & \vdots\\
x_{m1} & x_{m2} & \cdots & x_{mn}
\end{bmatrix}
{\begin{bmatrix}
w_{11} & w_{12} & \cdots & w_{1n}\\
w_{21} & w_{22} & \cdots & w_{2n}
\end{bmatrix}}^T
+\begin{bmatrix}
b_{11} & b_{21} \\
b_{12} & b_{22} \\
\vdots & \vdots \\
b_{1n} & b_{2n}
\end{bmatrix}
$$



### 线性回归的解析解推导过程

#### 问题设置

设有 $m$ 个样本，每个样本有 $n$ 个特征。将偏置 $b$ 并入参数向量，在 $X$ 末尾追加一列全 1：

$$
\tilde{X} = \begin{bmatrix} x_{11} & \cdots & x_{1n} & 1\\ \vdots & \ddots & \vdots & \vdots\\ x_{m1} & \cdots & x_{mn} & 1 \end{bmatrix}_{m\times(n+1)},\quad
\tilde{w} = \begin{bmatrix} w_1\\ \vdots\\ w_n\\ b \end{bmatrix}_{(n+1)\times 1}
$$

预测值为 $\hat{y} = \tilde{X}\tilde{w}$，目标是求使损失最小的 $\tilde{w}$（下文简记为 $w$，$X$ 即增广矩阵）。

#### 损失函数

均方误差（MSE）：

$$
L(w) = \frac{1}{2m}\|Xw - y\|^2 = \frac{1}{2m}(Xw-y)^T(Xw-y)
$$

> **为什么 $\|v\|^2 = v^Tv$？**
>
> 设残差向量 $v = Xw - y \in \mathbb{R}^m$，其 L2 范数的平方按定义是各分量平方和：
>
> $$\|v\|^2 = v_1^2 + v_2^2 + \cdots + v_m^2 = \sum_{i=1}^{m} v_i^2$$
>
> 而列向量与自身的转置相乘（内积）恰好也是各分量平方和：
>
> $$v^T v = \begin{bmatrix}v_1 & v_2 & \cdots & v_m\end{bmatrix} \begin{bmatrix}v_1\\v_2\\\vdots\\v_m\end{bmatrix} = v_1^2 + v_2^2 + \cdots + v_m^2$$
>
> 两者完全相等，因此 $\|v\|^2 = v^Tv$。

展开：

$$
L(w) = \frac{1}{2m}\left(w^TX^TXw - 2y^TXw + y^Ty\right)
$$

#### 对 $w$ 求梯度

利用矩阵微分公式 $\dfrac{\partial}{\partial w}(w^TAw)=2Aw$（$A$ 对称）和 $\dfrac{\partial}{\partial w}(b^Tw)=b$：

$$
\frac{\partial L}{\partial w} = \frac{1}{m}\left(X^TXw - X^Ty\right)
$$

#### 令梯度为零

> 求解函数极值的思路

$$
X^TXw = X^Ty
$$

这称为**正规方程（Normal Equation）**。

#### 求解

当 $X^TX$ 可逆时（列满秩），唯一解为：

> 列满秩的直觉理解：每一列都提供了独立的新信息；每一列可以看作一个特征；意味着没有任何一列可以被其他列的线性组合表示出来；每个特征均不可被替代

$$
\boxed{w^* = (X^TX)^{-1}X^Ty}
$$

$(X^TX)^{-1}X^T$ 称为 $X$ 的**伪逆（Moore-Penrose pseudoinverse）**，记作 $X^+$。

> 普通逆：要求矩阵是方阵且可逆；但这里的X 是一个 m*n的矩阵，样本数 m 远大于 特征数 n。
> y=Xw是一个超定方程组，m 个方程 n 个未知数，方程比未知数多，通常无精确解。
> 伪逆的思路是退而求次优，找使得残差最小的w

#### 验证为极小值

对梯度再关于 $w^T$ 求导，得到 Hessian 矩阵：

$$
H = \frac{\partial^2 L}{\partial w\, \partial w^T}
= \frac{\partial}{\partial w^T}\left[\frac{1}{m}(X^TXw - X^Ty)\right]
$$

$X^Ty$ 与 $w$ 无关，求导为 0；利用公式 $\dfrac{\partial (Aw)}{\partial w^T} = A$：

$$
H = \frac{1}{m} X^TX
$$

**$H$ 为何半正定？** 对任意非零向量 $z$：

$$
z^T H z = \frac{1}{m} z^T X^T X z = \frac{1}{m}(Xz)^T(Xz) = \frac{1}{m}\|Xz\|^2 \geq 0
$$

范数平方恒非负，故 $H \succeq 0$（半正定）。

**$X$ 列满秩时 $H$ 正定：** $z \neq 0$ 时 $Xz \neq 0$，故 $\|Xz\|^2 > 0$，即 $z^THz > 0$，$H \succ 0$（正定）。

Hessian 正定 $\Rightarrow$ $L(w)$ 严格凸 $\Rightarrow$ 梯度为零的驻点是**唯一全局最小值点**。

> **半正定的直观理解：** $z^TAz \geq 0$ 对所有 $z$ 成立，意味着二次型 $f(z)=z^TAz$ 定义的曲面"只有谷底和平坦，绝不下陷"——正定是碗形（唯一最低点），半正定是槽形（某方向平坦但不下坡），不定才有鞍点。
> 对 $X^TX$ 而言：$z^TX^TXz = \|Xz\|^2$，长度平方恒 $\geq 0$，所以半正定是"平方"这件事本身保证的。

#### 小结

符号：$m$ 样本数，$n$ 特征数，$T$ 迭代轮数，$b$ mini-batch 大小

| 方法 | 公式 | 时间复杂度 | 空间复杂度 | 适用场景 |
|------|------|-----------|-----------|----------|
| 解析解（正规方程） | $w^* = (X^TX)^{-1}X^Ty$ | $O(mn^2 + n^3)$ | $O(mn + n^2)$ | $n$ 较小、$X^TX$ 可逆 |
| 梯度下降（全批量） | $w \leftarrow w - \dfrac{\alpha}{m}X^T(Xw-y)$ | $O(T \cdot mn)$ | $O(mn)$ | $n$ 大、数据能放入内存 |
| 随机梯度下降（mini-batch） | 同上，每次用 $b$ 个样本估计梯度 | $O(T \cdot bn)$ | $O(bn)$ | 数据量极大、在线学习 |

**时间复杂度说明：**
- 解析解的瓶颈：计算 $X^TX$ 需 $O(mn^2)$，对其求逆需 $O(n^3)$；$n$ 达到万级时 $n^3$ 不可接受
- 梯度下降每轮只做矩阵-向量乘法 $O(mn)$，用迭代换掉了矩阵求逆；$T$ 取决于学习率与数据条件数

**空间复杂度说明：**
- 解析解需同时驻留整个 $X$（$O(mn)$）和 $n\times n$ 矩阵（$O(n^2)$）
- mini-batch 每次只需读入 $b$ 行，空间降至 $O(bn)$，适合数据集远大于内存的场景
