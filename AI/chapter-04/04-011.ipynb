## softmax

```python
import torch
```

```python
def softmax(X):
    X_exp = torch.exp(X)
    partition = X_exp.sum(1, keepdim=True)
    return X_exp / partition
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

fm.fontManager.addfont('/System/Library/Fonts/Hiragino Sans GB.ttc')
plt.rcParams['font.family'] = 'Hiragino Sans GB'
plt.rcParams['axes.unicode_minus'] = False

# 3 个类别的 logits 示例
logits = torch.tensor([[0.3, 2.9, 4.0]])
probs  = softmax(logits).squeeze().numpy()
logits_np = logits.squeeze().numpy()
print(logits_np, probs)
print(type(logits_np), type(probs))

classes = ['类别 0', '类别 1', '类别 2']
x = np.arange(len(classes))

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].bar(x, logits_np, color='steelblue')
axes[0].set_title('输入 logits')
axes[0].set_xticks(x)
axes[0].set_xticklabels(classes)
axes[0].set_ylabel('值')
axes[0].grid(True, alpha=0.3, axis='y')

axes[1].bar(x, probs, color='coral')
axes[1].set_title('softmax 输出（概率）')
axes[1].set_xticks(x)
axes[1].set_xticklabels(classes)
axes[1].set_ylabel('概率')
axes[1].set_ylim(0, 1)
axes[1].grid(True, alpha=0.3, axis='y')
for i, v in enumerate(probs):
    axes[1].text(i, v + 0.02, f'{v:.3f}', ha='center')

plt.suptitle('softmax: logits -> 概率分布（和为 1）')
plt.tight_layout()
plt.show()
```
